```python
import json, joblib, lightgbm
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. load our model
art = joblib.load("../models/lgbm_model_v3.joblib")
model, cols, thr = art["model"], art["columns"], art["threshold"]

# 2. saveing the model as plain text 
model.booster_.save_model("../models/lgbm_v3.txt")

# 3. save column order + threshold as JSON
with open("../models/meta_v3.json", "w") as f:
    json.dump({"columns": cols, "threshold": float(thr)}, f, indent=2)
```

```python
df = pd.read_csv("../data/processed/startup_data_final.csv")
x, y = df[cols], df["label"]
_, x_test, _, _ = train_test_split(x, y, test_size=0.2, stratify=y, random_state=42)

booster = lightgbm.Booster(model_file="../models/lgbm_v3.txt")

print(np.allclose(booster.predict(x_test), model.predict_proba(x_test)[:, 1]))
```
